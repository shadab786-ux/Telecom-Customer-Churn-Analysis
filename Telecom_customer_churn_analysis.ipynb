# Using IQR Method:

```python
def find_outliers(column):
    
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]
    
    print("Column:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Limit:", lower)
    print("Upper Limit:", upper)
    print("Number of Outliers:", len(outliers))
    
    return outliers
monthly_outliers = find_outliers("MonthlyCharges") 
print("-------------------------------------------")
total_outliers = find_outliers("TotalCharges")
print("-------------------------------------------")
enure_outliers = find_outliers("tenure")
```

# Identify Churn Factors:

```python
factors = [
    "Contract",
    "InternetService",
    "PaymentMethod",
    "TechSupport",
    "OnlineSecurity",
    "PaperlessBilling",
    "Partner",
    "Dependents"
]

for col in factors:
    
    print("\n" + "="*50)
    print(col)
    
    result = pd.crosstab(
        df[col],
        df["Churn"],
        normalize="index"
    ) * 100
    
    print(result)
```

# Automatically Find Highest Churn Category:

```python
results = []

for col in factors:
    
    result = pd.crosstab(
        df[col],
        df["Churn"],
        normalize="index"
    ) * 100
    
    if "Yes" in result.columns:
        
        category = result["Yes"].idxmax()
        rate = result["Yes"].max()
        
        results.append({
            "Feature": col,
            "Category": category,
            "Churn Rate (%)": round(rate, 2)
        })

churn_factor_summary = pd.DataFrame(results)

churn_factor_summary.sort_values(
    "Churn Rate (%)",
    ascending=False
)
```

# Final KPI Table

```python
total_customers = len(df)

churned_customers = (df["Churn"] == "Yes").sum()

active_customers = (df["Churn"] == "No").sum()

churn_rate = churned_customers / total_customers * 100

avg_tenure = df["tenure"].mean()

avg_monthly = df["MonthlyCharges"].mean()

avg_total = df["TotalCharges"].mean()

summary = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Churned Customers",
        "Active Customers",
        "Churn Rate",
        "Average Tenure",
        "Average Monthly Charges",
        "Average Total Charges"
    ],
    "Value": [
        total_customers,
        churned_customers,
        active_customers,
        round(churn_rate, 2),
        round(avg_tenure, 2),
        round(avg_monthly, 2),
        round(avg_total, 2)
    ]
})

summary
```
